```python
enrolment = [0.62, 0.88, 0.94, 0.67, 0.93, 0.95]
zones = ['North West', 'North Central', 'South West',
         'North East', 'South East', 'South South']

print(len(enrolment))
print(enrolment[0])
print(enrolment[-1])
print(enrolment[1:3])
```

```python
rates ={}
for z, e in zip(zones, enrolment):
  rates[z] = e

print(rates['North West'])
```

```python
for zone, rate in rates.items():
    print(f"{zone:15s} {rate:.0%}")
```

```python
def classify(rate):
  """Return a text label for an enrolment rate."""
  if rate >= 0.90:
     return "High"
  elif rate >= 0.75:
    return "Moderate"
  else:
      return "Low"

  for zone, rate in rates.items():
      print(f"{zone:15s} {rate:.0%}  {classify(rate)}")
```

```python
def describe(zone, rate, decimals=1):
    pct = round(rate * 100, decimals)
    return f"{zone}: {pct}% enrolled ({classify(rate)})"

print(describe('Kano', 0.61))
print(describe('Kano', 0.61, decimals=2))
```
